```python
# Program to input a 3x3 matrix and calculate sum of each row and column

# Input the 3x3 matrix
matrix = []
print("Enter elements of 3x3 matrix row-wise:")
for i in range(3):
    row = []
    for j in range(3):
        row.append(int(input(f"Enter element [{i}][{j}]: ")))
    matrix.append(row)

# Display the matrix
print("\nGiven Matrix:")
for row in matrix:
    print(row)

# Calculate and display sum of each row
print("\nSum of each row:")
for i in range(3):
    row_sum = sum(matrix[i])
    print(f"Row {i+1}: {row_sum}")

# Calculate and display sum of each column
print("\nSum of each column:")
for j in range(3):
    col_sum = sum(matrix[i][j] for i in range(3))
    print(f"Column {j+1}: {col_sum}")
```
