```python
import os
import datetime
import platform
import subprocess
import threading
import time
from pynput import keyboard

# ----------------------------------------------
# Action functions – each shortcut triggers one of these
# ----------------------------------------------

def action_print_message():
    """
    Print a timestamped message to the console.
    This is triggered by Ctrl+Alt+D.
    """
    print("Shortcut Ctrl+Alt+D activated – printing current time...")
    print(f"Current time: {datetime.datetime.now()}")

def action_open_file():
    """
    Create/append to a text file and open it with the default editor.
    Triggered by Ctrl+Alt+T.
    """
    file_path = "example_shortcut_file.txt"
    content = f"File opened via Ctrl+Alt+T at {datetime.datetime.now()}\n"

    # Append content to the file (creates it if it doesn't exist)
    with open(file_path, "a", encoding="utf-8") as f:
        f.write(content)

    print(f"File '{file_path}' updated.")

    # Open the file with the system's default application
    if platform.system() == "Windows":
        os.startfile(file_path)
    else:
        subprocess.call(["xdg-open", file_path])

def action_show_directory():
    """List all files in the current working directory. Triggered by Ctrl+Alt+P."""
    print(" Contents of current directory:")
    for item in os.listdir("."):
        print("  -", item)

def action_system_info():
    """Display basic system information. Triggered by Ctrl+Alt+I."""
    print(" System information:")
    print(f"  OS: {platform.system()}")
    print(f"  Release: {platform.release()}")
    print(f"  Machine: {platform.machine()}")

def action_exit():
    """Stop the listener. This is a special shortcut (Ctrl+Alt+Q)."""
    print(" Exiting listener...")
    # We'll raise an exception to break out of the main loop (handled later)
    raise KeyboardInterrupt

# ----------------------------------------------
# Dictionary mapping hotkeys to their corresponding action
# Format: '<ctrl>+<alt>+d' means hold Ctrl and Alt and press d
# ----------------------------------------------
shortcuts = {
    '<ctrl>+<alt>+d': action_print_message,
    '<ctrl>+<alt>+t': action_open_file,
    '<ctrl>+<alt>+p': action_show_directory,
    '<ctrl>+<alt>+i': action_system_info,
    '<ctrl>+<alt>+q': action_exit,   # Exit the program
}

# ----------------------------------------------
# Listener setup using GlobalHotKeys
# ----------------------------------------------

def on_activate(shortcut_name):
    """
    Callback invoked when a registered hotkey is pressed.
    It finds the corresponding action and executes it.
    """
    print(f"\n Hotkey activated: {shortcut_name}")
    if shortcut_name in shortcuts:
        shortcuts[shortcut_name]()
    else:
        print("Unknown shortcut.")

# Create the GlobalHotKeys listener
listener = keyboard.GlobalHotKeys({
    hotkey: (lambda name=hotkey: on_activate(name))
    for hotkey in shortcuts
})

# Start the listener in a background thread so the notebook remains interactive
listener_thread = threading.Thread(target=listener.start, daemon=True)
listener_thread.start()

print(" Listening for hotkeys...")
print("Available shortcuts:")
for key in shortcuts:
    print(f"  - {key}")

print("\nPress Ctrl+Alt+Q to stop the listener, or interrupt the cell.")
```
