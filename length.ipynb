```python
# Python program to find words with the most letters in a sentence and display their length


sentence= input("please enter your sentence:")
#Split to words
words=sentence.split()


if not words:
    print("There is no word in the sentence!")
else:
    max_length = max(len(word) for word in words) #Find Max Length
    
    longest_words=[word for word in words if len(word) == max_length]
    
    print(f"Longest word(s): {', '.join(longest_words)}")
    print(f"Length: {max_length}")
    
    
```
