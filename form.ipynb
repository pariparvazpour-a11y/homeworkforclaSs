```python
from flask import Flask, request, render_template_string, redirect, url_for, flash
app = Flask(__name__)
app.secret_key = "secret_key_for_flash"

records = []

HTML_PAGE = """
<!doctype html>
<html lang="fa" dir="rtl">
<head>
  <meta charset="utf-8" />
  <meta name="viewport" content="width=device-width, initial-scale=1" />
  <title>ثبت علاقه‌مندی‌ها</title>
  <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/css/bootstrap.rtl.min.css" rel="stylesheet">
</head>
<body class="bg-light">
  <div class="container py-4">
    <h2 class="mb-3">فرم ثبت اطلاعات</h2>

    {% with messages = get_flashed_messages(with_categories=true) %}
      {% if messages %}
        {% for category, message in messages %}
          <div class="alert alert-{{category}} alert-dismissible fade show" role="alert">
            {{message}}
            <button type="button" class="btn-close" data-bs-dismiss="alert" aria-label="Close"></button>
          </div>
        {% endfor %}
      {% endif %}
    {% endwith %}

    <div class="card shadow-sm mb-4">
      <div class="card-body">
        <form method="POST">
          <div class="row g-3">
            <div class="col-md-4">
              <label class="form-label">نام</label>
              <input type="text" name="name" class="form-control" placeholder="مثال: محمد" required>
            </div>

            <div class="col-md-2">
              <label class="form-label">سن</label>
              <input type="number" name="age" class="form-control" placeholder="مثال: 20" required>
            </div>

            <div class="col-md-6">
              <label class="form-label">علاقه‌مندی</label>
              <input type="text" name="interest" class="form-control" placeholder="مثال: ورزش، موسیقی، برنامه‌نویسی">
            </div>

            <div class="col-md-6">
              <label class="form-label">میزان علاقه</label>
              <select name="level" class="form-select" required>
                {% for i in range(1, 11) %}
                  <option value="{{i}}">{{i}}</option>
                {% endfor %}
              </select>
            </div>

            <div class="col-md-6">
              <label class="form-label">توضیحات اضافی</label>
              <textarea name="description" class="form-control" rows="2" placeholder="اختیاری..."></textarea>
            </div>
          </div>

          <div class="mt-3">
            <button class="btn btn-primary" type="submit">Submit</button>
          </div>
        </form>
      </div>
    </div>

    <div class="card shadow-sm">
      <div class="card-header bg-dark text-white">
        <strong>لیست رکوردها</strong>
      </div>
      <div class="card-body">
        {% if records %}
          <div class="table-responsive">
            <table class="table table-striped table-bordered align-middle">
              <thead class="table-secondary">
                <tr>
                  <th>نام</th>
                  <th>سن</th>
                  <th>علاقه‌مندی</th>
                  <th>میزان علاقه</th>
                  <th>توضیحات</th>
                </tr>
              </thead>
              <tbody>
                {% for r in records %}
                  <tr>
                    <td>{{r.name}}</td>
                    <td>{{r.age}}</td>
                    <td>{{r.interest}}</td>
                    <td><span class="badge text-bg-info">{{r.level}}/10</span></td>
                    <td>{{r.description}}</td>
                  </tr>
                {% endfor %}
              </tbody>
            </table>
          </div>
        {% else %}
          <p class="text-muted mb-0">هنوز رکوردی ثبت نشده است.</p>
        {% endif %}
      </div>
    </div>
  </div>

  <script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/js/bootstrap.bundle.min.js"></script>
</body>
</html>
"""

@app.route("/", methods=["GET", "POST"])
def index():
    if request.method == "POST":
        name = (request.form.get("name") or "").strip()
        age = (request.form.get("age") or "").strip()
        interest = (request.form.get("interest") or "").strip()
        level = (request.form.get("level") or "").strip()
        description = (request.form.get("description") or "").strip()

        if not name:
            flash("فیلد نام نمی‌تواند خالی باشد.", "danger")
            return redirect(url_for("index"))

        if not age.isdigit():
            flash("سن باید عدد صحیح باشد.", "danger")
            return redirect(url_for("index"))

        # level از select می‌آید (1..10)
        if not level.isdigit() or not (1 <= int(level) <= 10):
            flash("میزان علاقه باید بین ۱ تا ۱۰ باشد.", "danger")
            return redirect(url_for("index"))

        records.append({
            "name": name,
            "age": int(age),
            "interest": interest,
            "level": int(level),
            "description": description
        })
        flash("اطلاعات با موفقیت ثبت شد.", "success")
        return redirect(url_for("index"))

    return render_template_string(HTML_PAGE, records=records)

# اجرا
app.run(debug=True, port=5000, use_reloader=False)
```
