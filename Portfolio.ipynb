# app.py

```python
%%writefile portfolio/app.py

from flask import Flask, render_template

app = Flask(__name__)


# Personal information
PERSONAL_INFO = {
    "name": " Farzaneh",
    "job": "AI engineer",
    "bio": (
        "I am a computer engineer who is interested in website design"
    ),
    "email": "farzaneh@example.com",
    "github": "https://github.com/pariparvazpour-a11y"
}


# Skills list
SKILLS = [
    "Python",
    "Flask",
    "HTML",
    "CSS",
    "JavaScript",
    "Git and GitHub",
    "SQL"
]


# Projects list
PROJECTS = [
    {
        "title": "Online Shop",
        "description": (
            "A simple online shop built with Flask and SQLite."
        ),
        "github": "https://github.com/your-username/shop-project"
    },
    {
        "title": "Task Manager",
        "description": (
            "A task management application for creating and managing "
            "daily tasks."
        ),
        "github": "https://github.com/your-username/task-manager"
    },
    {
        "title": "Personal Blog",
        "description": (
            "A simple blog for publishing and organizing personal posts."
        ),
        "github": "https://github.com/your-username/personal-blog"
    }
]


@app.route("/")
def home():
    return render_template(
        "home.html",
        person=PERSONAL_INFO
    )


@app.route("/skills")
def skills():
    return render_template(
        "skills.html",
        person=PERSONAL_INFO,
        skills=SKILLS
    )


@app.route("/projects")
def projects():
    return render_template(
        "projects.html",
        person=PERSONAL_INFO,
        projects=PROJECTS
    )


if __name__ == "__main__":
    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False
    )
```
